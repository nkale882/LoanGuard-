# Loan approval governance demo

This project demonstrates how Knowlytix governance components can support a lending workflow. The policy and 12 applications are synthetic. A route marked eligible is only a handoff for human underwriting; it is never an approval.

Run notebooks 00 through 03 in order. Notebook 03 makes live Groq requests and can incur provider usage. The local notebooks run without an LLM key.


## Project map

- `loan_policy.md`: the source policy ingested into GMS.
- `synthetic_applications.csv`: invented application rows and illustrative expected routes.
- `loanguard.py`: reusable pipeline functions and a command-line walkthrough.
- `loan_store/`: local GMS cache, created or loaded by Knowlytix.
- `out/`: audit artifacts and a factual demo summary generated by the pipeline.

## Run

Select the Python environment with Knowlytix installed. Start at notebook 01, then open notebook 02 for group-level review and notebook 03 for optional Groq-backed claim extraction. From a terminal in the project root, `python loanguard.py` runs the deterministic workflow. Add `--rebuild` to re-ingest and train the local store.


## Interpretation

Knowlytix provides knowledge ingestion, typed claim verification, policy gating, governance tracing and provenance. This demo uses those components for an educational pre-screen. The sample is too small to validate lending policy, fairness, model performance, or regulatory compliance.
